```python
df_customer = pd.json_normalize(df["customer"])
```

```python
df_phone = pd.json_normalize(df["phone"])
```

```python
df_internet = pd.json_normalize(df["internet"])
```

```python
df_account = pd.json_normalize(df["account"])
```

```python
df_base = df[["customerID", "Churn"]]
```

```python
df_normalizado = pd.concat([df_base, df_customer, df_phone, df_internet, df_account],axis=1)
```

```python
df_normalizado.head()
```

```python
df_normalizado.dtypes
```

```python
df_normalizado["Charges.Total"] = pd.to_numeric(df_normalizado["Charges.Total"], errors="coerce"
)
```

```python
df_normalizado.dtypes
```

```python
df_normalizado.isnull().sum()
```

```python
df_normalizado["customerID"].duplicated().sum()
```

```python
df_normalizado.duplicated().sum()
```

```python
df_normalizado["Cuentas_Diarias"] = df_normalizado["Charges.Monthly"] / 30
```

```python
df_normalizado.head()
```

```python
df_normalizado.describe()
```

```python
df_normalizado["Churn"].value_counts()
```

```python
df_normalizado["Churn"] = df_normalizado["Churn"].replace("", pd.NA)
```

```python
df_normalizado["Churn"].value_counts(dropna=False)
```

```python
df_churn = df_normalizado[df_normalizado["Churn"].notna()]
```

```python
df_churn["Churn"].value_counts()
```

```python
import matplotlib.pyplot as plt

```

```python
df_churn["Churn"].value_counts().plot(kind="bar")
plt.title("Distribución de Churn")
plt.xlabel("Churn")
plt.ylabel("Cantidad de clientes")
plt.show()
```

```python
df_churn["Churn"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Proporción de clientes con y sin Churn")
plt.ylabel("")
plt.show()
```

```python
df_churn["gender"].value_counts()
```

```python
df_churn.groupby("gender")["Churn"].value_counts()
```

```python
df_churn.groupby("gender")["Churn"].value_counts().unstack().plot(kind="bar")
plt.title("Churn según género")
plt.xlabel("Género")
plt.ylabel("Cantidad de clientes")
plt.show()
```

```python
df_churn.groupby("Contract")["Churn"].value_counts().unstack().plot(kind="bar")
plt.title("Churn según tipo de contrato")
plt.xlabel("Tipo de contrato")
plt.ylabel("Cantidad de clientes")
plt.show()
```

```python
df_churn.groupby("PaymentMethod")["Churn"].value_counts().unstack().plot(kind="bar")
plt.title("Churn según método de pago")
plt.xlabel("Método de pago")
plt.ylabel("Cantidad de clientes")
plt.show()
```

```python
df_churn.boxplot(column="Charges.Total", by="Churn")
plt.title("Total gastado según Churn")
plt.suptitle("")  # quita el título automático feo
plt.xlabel("Churn")
plt.ylabel("Total gastado")
plt.show()
```

```python
df_churn.boxplot(column="tenure", by="Churn")
plt.title("Tiempo de contrato según Churn")
plt.suptitle("")   # quita el título automático
plt.xlabel("Churn")
plt.ylabel("Meses de contrato")
plt.show()
```

##INTRODUCCION

El objetivo de este análisis es comprender el comportamiento de evasión de clientes en la empresa TelecomX.  
El churn ocurre cuando un cliente decide cancelar el servicio, lo que representa una pérdida directa de ingresos y de clientes para la empresa.

A través de este análisis se busca:
- Identificar la proporción de clientes que cancelan y los que permanecen.
- Analizar cómo diferentes características influyen en la evasión.
- Encontrar patrones que permitan diseñar estrategias para reducir el churn.


##LIMPIEZA Y TRATAMIENTO DE LOS DATOS


Para preparar los datos se realizaron los siguientes pasos:

- Se cargó el archivo en formato JSON utilizando la librería pandas.
- Se identificaron columnas que contenían información anidada y se normalizaron.
- Se unificaron todas las tablas en un solo DataFrame.
- Se convirtió la columna Charges.Total a tipo numérico.
- Se detectaron valores vacíos en la columna Churn que estaban representados como texto vacío.
- Estos valores fueron reemplazados por valores nulos reales `NaN` y posteriormente eliminados.
- Se creó el DataFrame df_churn que contiene únicamente los registros válidos para el análisis.
- Se verificó que no existieran registros duplicados.

Con estos pasos se aseguró que el conjunto de datos estuviera limpio, consistente y listo para el análisis exploratorio.


##ANALISIS EXPLORATORIO DE LOS DATOS

### 1. Distribución de la variable Churn
  
Aproximadamente el 73.5% de los clientes permanecen activos, mientras que el 26.5% cancelan el servicio.

### 2. Churn según género

Al analizar la evasión según el género, se observa que tanto hombres como mujeres presentan comportamientos muy similares.  
La proporción de clientes que cancelan es casi la misma en ambos grupos.

Esto indica que el género no es un factor determinante fuerte para explicar el churn.

image.png

### 3. Churn según tipo de contrato

El análisis por tipo de contrato muestra una diferencia clara:

- Los clientes con contrato **mes a mes** presentan la mayor cantidad de cancelaciones.
- Los clientes con contrato de **un año** tienen menos churn.
- Los clientes con contrato de **dos años** casi no presentan cancelaciones.

Esto indica que mientras más largo es el compromiso contractual, menor es la probabilidad de churn.

image.png

### 4. Churn según método de pago

El análisis por método de pago muestra que:

- Los clientes que pagan mediante **electronic check** presentan una mayor cantidad de cancelaciones.
- Los métodos de pago automáticos (tarjeta de crédito o débito automático) muestran menor churn.

Esto sugiere que los clientes con pagos automatizados tienen mayor permanencia.

image.png

### 5. Churn según total gastado

A partir del boxplot se observa que:

- Los clientes que no cancelan presentan un mayor gasto total acumulado.
- Los clientes que cancelan tienden a haber gastado menos.

Esto indica que los clientes con mayor inversión en el servicio tienden a permanecer más tiempo.

image.png

### 6. Churn según tiempo de contrato

El boxplot muestra que:

- Los clientes que no cancelan tienen un tiempo de contrato mucho mayor.
- Los clientes que cancelan generalmente tienen pocos meses de permanencia.

El tiempo de contrato es uno de los factores más claros asociados con el churn.

image.png

##CONCLUSIONES

Del análisis realizado se concluye que:

- El tiempo de contrato es uno de los factores más importantes para explicar el churn.
- Los contratos mensuales presentan el mayor riesgo de cancelación.
- Los clientes con menor gasto acumulado tienen mayor tendencia a cancelar.
- El método de pago influye en la evasión, siendo el *electronic check* el que muestra más churn.
- El género no representa una diferencia significativa en el comportamiento de cancelación.
